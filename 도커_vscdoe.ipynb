# Docker 컨테이너 VS Code 연결

Docker 컨테이너를 터미널이 아니라 VS Code 작업 환경으로 여는 방법입니다.

필요한 것:
- Docker 실행 상태
- VS Code 설치
- VS Code 확장: `Dev Containers`


## 1. 컨테이너 실행

VS Code로 연결할 컨테이너는 실행 중이어야 합니다.

### Linux/macOS 또는 WSL
```bash
docker run -d --name my_python_container \
  -p 8000:8000 \
  -v "$PWD":/home \
  -w /home \
  python:3.9 sleep infinity
```

### Windows PowerShell
```powershell
docker run -d --name my_python_container `
  -p 8000:8000 `
  -v "${PWD}:/home" `
  -w /home `
  python:3.9 sleep infinity
```


## 2. 실행 확인

```bash
docker ps
```

`my_python_container`가 보이면 정상입니다.

컨테이너에 터미널로 들어가려면 아래 명령을 사용합니다.

```bash
docker exec -it my_python_container bash
```


## 3. VS Code에서 연결

1. VS Code를 실행합니다.
2. 왼쪽 Extensions에서 `Dev Containers` 확장을 설치합니다.
3. `Ctrl + Shift + P`를 누릅니다.
4. `Dev Containers: Attach to Running Container...`를 선택합니다.
5. `my_python_container`를 선택합니다.

새 VS Code 창이 열리면 컨테이너에 연결된 상태입니다.


## 4. 폴더 열기

컨테이너에 연결된 VS Code 창에서 작업 폴더를 엽니다.

1. `File` -> `Open Folder...`
2. `/home` 입력
3. `OK` 선택

`docker run`에서 현재 폴더를 `/home`에 연결했기 때문에, `/home`에서 실습 파일을 볼 수 있습니다.


## 5. 간단 확인

VS Code 메뉴에서 `Terminal` -> `New Terminal`을 선택한 뒤 확인합니다.

```bash
pwd
python --version
ls
```

`pwd`가 `/home`으로 나오고 실습 파일이 보이면 연결이 완료된 것입니다.


## 안 될 때 확인

- `docker ps`에서 컨테이너가 실행 중인지 확인합니다.
- 컨테이너가 꺼져 있으면 `docker start my_python_container`를 실행합니다.
- VS Code에 컨테이너 목록이 보이지 않으면 Docker가 실행 중인지 확인합니다.
- 같은 이름의 컨테이너를 다시 만들려면 기존 컨테이너를 삭제합니다.

```bash
docker stop my_python_container
docker rm my_python_container
```
