## 1. Streamlit 시작하기

### 1.1 Streamlit이란?

Streamlit은 **Python 코드만으로 웹앱을 만드는 오픈소스 프레임워크**입니다.  
HTML, CSS, JavaScript를 몰라도 입력·출력·차트가 있는 화면을 만들 수 있어서 **AI 모델 데모, 대시보드, 프로토타입** 제작에 많이 쓰입니다.

- **코드 중심 UI**: `st.title()`, `st.button()` 같은 함수 호출이 곧 화면 구성
- **즉시 피드백**: 코드를 저장하면 브라우저에 바로 반영
- **시각화 통합**: 표와 차트를 함수 한 줄로 출력
- **쉬운 배포**: GitHub 저장소만 있으면 Streamlit Community Cloud에 무료 배포

### 1.2 웹앱 프레임워크 비교

AI 모델을 서비스하려면 사용자의 입력을 받아 모델 결과를 보여주는 **웹 프레임워크**가 필요합니다.

| 구분 | Streamlit | Gradio | FastAPI / Flask |
|------|------------|---------|-----------------|
| 주 목적 | 화면이 있는 데이터 앱 | 모델 체험 데모 | API 서버 |
| 특징 | Python 스크립트가 곧 웹앱 | 함수 하나에 UI를 자동 생성 | 화면 없이 요청/응답(REST)만 처리 |
| 권장 활용 | 내부 프로토타입, 대시보드 | AI 데모 | 상용 서비스 백엔드 |

- Streamlit → 빠른 프로토타입, 대시보드 (이번 강의)
- FastAPI → 다른 프로그램이 호출하는 모델 API 서버 (다음 강의 4-4)

### 1.3 첫 앱 실행

1. 실습 폴더를 만들고 그 안으로 이동합니다. **이번 강의의 모든 명령은 이 폴더 안에서 실행합니다.**

```bash
mkdir ai-prototype   # 실습 폴더 생성
cd ai-prototype      # 실습 폴더로 이동
```

2. `app.py` 파일을 만들고 아래 코드를 입력합니다.

```python
import streamlit as st

st.set_page_config(page_title="AI Prototype", layout="wide")  # 브라우저 탭 제목, 넓은 화면 레이아웃
st.title("AI Prototype: Streamlit")                           # 페이지 제목
st.write("Streamlit 앱이 정상적으로 실행되고 있습니다.")         # 일반 텍스트 출력
```

3. 터미널에서 실행합니다.

```bash
streamlit run app.py
```

4. 브라우저에서 `localhost:8501`이 열리는지 확인합니다.
5. `st.write()` 안의 문장을 바꾸고 저장한 뒤 브라우저 화면이 바뀌는지 확인합니다. (오른쪽 위에 `Rerun`이 보이면 클릭하거나 `Always rerun`을 선택합니다.)

#### 실행 원리: rerun

> Streamlit은 버튼 클릭, 입력값 변경, 새로고침이 일어날 때마다 **스크립트 전체를 위에서 아래로 다시 실행(rerun)** 하고 그 결과를 화면에 그립니다.  
> 이 한 가지 원리가 뒤에서 배울 상태 관리(2장)와 캐싱(3장)이 필요한 이유입니다.

## 2. 기본 구성요소로 컴포넌트 카탈로그 만들기

이 장에서는 입력 위젯, 출력 요소, 레이아웃, 상태 관리를 배웁니다.  
**이 장의 예제 코드는 모두 `pages/01_components.py` 한 파일에 차례로 이어 붙여서** 자주 쓰는 컴포넌트를 모아 둔 "컴포넌트 카탈로그" 페이지를 완성합니다.

### 2.1 다중 페이지 구조 (`pages/`)

`app.py`와 같은 위치에 `pages/` 폴더를 만들고 그 안에 `.py` 파일을 넣으면, Streamlit이 자동으로 **사이드바에 페이지 메뉴**를 만듭니다.

```text
ai-prototype/
 ├─ app.py                  # 첫 화면 (메뉴 이름: app)
 └─ pages/
     └─ 01_components.py    # 메뉴 이름: components
```

> 파일 이름 앞의 숫자(`01_`, `02_`)는 메뉴 순서를 정하고, 메뉴에는 숫자를 뺀 이름이 표시됩니다.

```bash
mkdir pages   # pages 폴더 생성
```

### 2.2 입력 위젯

사용자에게 값을 입력받는 UI 요소입니다. 위젯 함수는 **사용자가 입력한 현재 값을 반환**합니다.

| 위젯 | 설명 | 예시 코드 |
|------|------|-----------|
| `st.text_input()` | 문자열 입력 | `name = st.text_input("이름 입력")` |
| `st.number_input()` | 숫자 입력 | `age = st.number_input("나이", 0, 120)` |
| `st.slider()` | 범위 선택 | `level = st.slider("난이도", 1, 10, 5)` |
| `st.selectbox()` | 드롭다운 선택 | `lang = st.selectbox("언어", ["Python", "R"])` |
| `st.checkbox()` | 체크박스 | `agree = st.checkbox("동의합니다")` |
| `st.button()` | 버튼 (누른 순간에만 `True`) | `if st.button("실행"):` |
| `st.file_uploader()` | 파일 업로드 | `file = st.file_uploader("CSV 업로드")` |

#### 실습: `pages/01_components.py` 만들기

파일을 새로 만들고 아래 코드를 입력합니다.

```python
import numpy as np                                      # 난수 생성 (2.3에서 사용)
import pandas as pd                                     # 표 데이터 (2.3에서 사용)
import streamlit as st

st.title("컴포넌트 카탈로그")                            # 페이지 제목

st.subheader("입력 위젯")                                # 소제목
name = st.text_input("이름 입력")                        # 문자열 입력
level = st.slider("숙련도", 1, 10, 5)                    # 1~10 중 선택, 기본값 5
lang = st.selectbox("언어 선택", ["Python", "Java", "C++"])  # 드롭다운 선택
if st.button("확인"):                                    # 버튼을 누른 그 실행에서만 True
    st.success(f"{name}님 / 숙련도 {level} / {lang}")    # 초록색 성공 메시지
```

저장 후 사이드바에서 **components**를 클릭해 확인합니다.

### 2.3 출력 요소

| 출력 함수 | 용도 | 예시 |
|------------|------|------|
| `st.write()` | 텍스트, 변수, 표를 알아서 출력 | `st.write("값:", x)` |
| `st.json()` | 딕셔너리를 JSON 형태로 출력 | `st.json({"ver": 1.0})` |
| `st.dataframe()` | 스크롤·정렬 가능한 표 | `st.dataframe(df)` |
| `st.bar_chart()` / `st.line_chart()` | 막대 / 선 차트 | `st.bar_chart(df)` |
| `st.pyplot()` / `st.plotly_chart()` / `st.altair_chart()` | 다른 시각화 라이브러리로 만든 차트 출력 | `st.pyplot(fig)` |

#### 실습: `pages/01_components.py` 아래에 이어서 추가

```python
st.subheader("출력 요소")
st.write("입력한 이름:", name)                           # 텍스트와 변수를 함께 출력
st.json({"name": name, "level": level, "lang": lang})   # 딕셔너리를 JSON 형태로 출력
data = pd.DataFrame({
    "x": np.arange(1, 6),                               # 1~5
    "y": np.random.randint(10, 100, 5),                 # 10~99 난수 5개
})
st.bar_chart(data.set_index("x"))                       # x를 가로축으로 막대 차트
```

> 슬라이더를 움직일 때마다 막대 높이가 바뀝니다. 위젯이 바뀔 때마다 스크립트 전체가 **rerun**되어 난수를 새로 뽑기 때문입니다.

### 2.4 레이아웃

| 함수 | 용도 |
|------|------|
| `st.columns(n)` | 화면을 가로로 n칸 나누기 |
| `st.container()` | 여러 요소를 하나의 묶음으로 보여주기 |
| `st.sidebar` | 왼쪽 사이드바에 요소 배치 |

#### 실습: `pages/01_components.py` 아래에 이어서 추가

```python
st.subheader("레이아웃")
left, right = st.columns(2)                             # 화면을 좌우 2칸으로 나눔
with left:                                              # 왼쪽 칸에 들어갈 내용
    st.write("왼쪽: 표")
    st.dataframe(data)
with right:                                             # 오른쪽 칸에 들어갈 내용
    st.write("오른쪽: 선 차트")
    st.line_chart(data.set_index("x"))
with st.container(border=True):                         # 테두리가 있는 묶음 영역
    st.write("컨테이너 안의 내용은 한 덩어리로 보입니다.")
st.sidebar.title("컴포넌트 카탈로그")                    # 사이드바 제목
st.sidebar.info("위젯·레이아웃·상태 관리 예시 페이지입니다.")  # 사이드바 안내 박스
```

### 2.5 상태 관리 (`st.session_state`)

rerun될 때마다 스크립트의 일반 변수는 처음 값으로 돌아갑니다.  
`st.session_state`는 **접속한 사용자(세션)마다 따로 유지되는 저장 공간**으로, rerun이나 페이지 이동 후에도 값이 남아 있습니다.

#### 실습 ① 카운터: `pages/01_components.py` 아래에 이어서 추가

```python
st.subheader("Session State")
if "counter" not in st.session_state:                   # 처음 접속했을 때만
    st.session_state.counter = 0                        # 카운터를 0으로 준비
if st.button("카운터 증가"):                             # 버튼을 누르면
    st.session_state.counter += 1                       # 세션에 저장된 값을 1 증가
st.write("현재 카운트:", st.session_state.counter)       # rerun 후에도 값이 유지됨
```

## 3. 캐싱으로 rerun 비용 줄이기

### 3.1 왜 캐싱이 필요한가

Streamlit은 버튼 클릭, 입력값 변경, 새로고침 때마다 스크립트를 **처음부터 다시 실행**합니다.  
그때마다 큰 데이터나 모델을 다시 불러오면 앱이 매우 느려집니다.  
**캐시는 rerun을 막는 것이 아니라, 오래 걸리는 함수의 결과를 저장해 두었다가 다시 계산하지 않게 해주는 장치**입니다.

| 구분 | 캐시 대상 | 반환 방식 | 주 용도 |
|------|-----------|-----------|---------|
| `@st.cache_data` | 함수의 **결과값** | 매번 **복사본** 반환 | 데이터 로드, 전처리, 예측 결과 |
| `@st.cache_resource` | **객체 자체** | 항상 **같은 객체** 반환 | 모델, DB 연결 |

- 두 캐시 모두 **모든 사용자가 공유**합니다. 한 사람이 만든 캐시를 다른 사람도 그대로 사용합니다.
- 같은 함수를 **같은 인자**로 다시 호출하면 저장된 결과를 바로 돌려줍니다.

> **선택 기준 한 줄 요약**  
> 표·리스트 같은 결과값 → `cache_data`  
> 모델처럼 한 번 만들어 계속 같이 쓸 객체 → `cache_resource`

### 3.2 실습: 캐싱 실험실 (`pages/02_cache_lab.py`)

내용이 똑같은 함수 두 개를 하나는 캐시 없이, 하나는 `@st.cache_data`를 붙여서 3번씩 호출하고 걸린 시간을 비교합니다.

```python
import time
import pandas as pd
import streamlit as st

st.title("캐싱 실험실")


def load_data_plain():                                  # 캐시 없는 함수
    time.sleep(1)                                       # 1초 걸리는 데이터 로드를 흉내
    return pd.DataFrame({"A": range(1000), "B": range(1000, 2000)})


@st.cache_data                                          # 결과값을 캐시하는 함수
def load_data_cached():
    time.sleep(1)                                       # 내용은 위 함수와 똑같음
    return pd.DataFrame({"A": range(1000), "B": range(1000, 2000)})


rows = []                                               # 측정 결과를 모을 리스트
for run in [1, 2, 3]:                                   # 같은 호출을 3번 반복
    t0 = time.perf_counter()                            # 시작 시각
    load_data_plain()                                   # 캐시 없이 호출
    t_plain = time.perf_counter() - t0                  # 걸린 시간
    t0 = time.perf_counter()
    load_data_cached()                                  # 캐시를 사용해 호출
    t_cached = time.perf_counter() - t0
    rows.append({"호출": run, "캐시 없음": t_plain, "캐시 사용": t_cached})

df = pd.DataFrame(rows).set_index("호출")               # 호출 번호를 인덱스로
st.table(df.round(3))                                   # 소수점 3자리 표
st.line_chart(df)                                       # 두 방식의 시간을 선 차트로 비교
st.button("다시 실행")                                   # 누르면 스크립트가 다시 실행됨(rerun)
```

#### 결과 확인

| 상황 | 캐시 없음 | 캐시 사용 |
|------|-----------|-----------|
| 처음 열었을 때 | 1.0 / 1.0 / 1.0초 | **1.0** / 0.0 / 0.0초 |
| `다시 실행` 클릭 | 1.0 / 1.0 / 1.0초 | 0.0 / 0.0 / 0.0초 |
| 오른쪽 위 `⋮` → `Clear cache` 후 `다시 실행` | 1.0 / 1.0 / 1.0초 | **1.0** / 0.0 / 0.0초 |

캐시 없는 함수는 rerun할 때마다 매번 1초씩 걸리고, 캐시를 사용한 함수는 **맨 처음 한 번만** 실제로 실행됩니다.

#### 체크리스트

| 항목 | 완료 여부 |
|------|-----------|
| `pages/02_cache_lab.py` 실행 | ☐ |
| `다시 실행` 클릭 시 캐시 사용 쪽이 모두 0초 | ☐ |
| `Clear cache` 후 첫 호출이 다시 1초 | ☐ |

#### 복습문제

<details>
<summary>1. <code>@st.cache_data</code>와 <code>@st.cache_resource</code>의 차이는?</summary>

`cache_data`는 함수의 **결과값**을 저장하고 매번 복사본을 돌려줍니다.  
`cache_resource`는 모델·DB 연결 같은 **객체 자체**를 저장하고 항상 같은 객체를 돌려줍니다.
</details>

<details>
<summary>2. 캐시를 사용하면 rerun이 일어나지 않는가?</summary>

**아닙니다.** rerun은 그대로 일어나고, 캐시가 붙은 함수만 다시 계산하지 않고 저장된 결과를 돌려줍니다.
</details>

<details>
<summary>3. A 사용자가 접속해 캐시가 만들어진 뒤 B 사용자가 처음 접속하면, B는 1초를 기다려야 하는가?</summary>

**기다리지 않습니다.** 캐시는 모든 사용자가 공유하므로 B도 저장된 결과를 바로 받습니다.
</details>

### 4.1 모델 로딩·예측 (`src/model.py`)

3장에서 배운 두 캐시를 그대로 적용합니다. 모델은 `cache_resource`, 예측 결과는 `cache_data`입니다.

```python
import streamlit as st
import pandas as pd
from sklearn.pipeline import Pipeline                         # 여러 단계를 하나의 모델로 묶음
from sklearn.feature_extraction.text import CountVectorizer   # 문장을 단어 개수 벡터로 변환
from sklearn.linear_model import LogisticRegression           # 긍정/부정 분류기


@st.cache_resource                                            # 모델 객체는 한 번만 만들고 모든 사용자가 공유
def load_model():
    """6개 단어로 학습한 더미 감성 분류 모델"""
    X = ["good", "great", "awesome", "bad", "awful", "terrible"]  # 학습 문장
    y = [1, 1, 1, 0, 0, 0]                                    # 1 = 긍정, 0 = 부정
    model = Pipeline([
        ("vec", CountVectorizer()),                           # 1단계: 단어 개수 세기
        ("clf", LogisticRegression(max_iter=200)),            # 2단계: 로지스틱 회귀 분류
    ])
    model.fit(X, y)                                           # 학습 (처음 호출될 때 한 번만 실행)
    return model


@st.cache_data                                                # 같은 입력이면 예측 결과를 재사용
def predict_texts(texts):
    """문장 리스트를 받아 예측 결과 표를 반환"""
    model = load_model()                                      # 캐시된 모델 가져오기
    preds = model.predict(texts)                              # 예측 라벨 (0 또는 1)
    probs = model.predict_proba(texts)[:, 1]                  # 긍정(1)일 확률
    return pd.DataFrame({"text": texts, "label": preds, "prob": probs})
```

> 실습용 더미 모델이라 학습한 6개 단어만 알고, 그 밖의 단어(한국어 포함)는 긍정 확률이 0.5로 나옵니다.

### 4.2 로그 저장·읽기 (`src/utils.py`)

예측 결과를 `data/log.csv`에 쌓아 두면 5장에서 시각화할 수 있습니다. 여러 페이지에서 함께 쓸 함수이므로 `src/utils.py`에 둡니다.

```python
from pathlib import Path
import pandas as pd

LOG_PATH = Path("data/log.csv")                         # 예측 로그 저장 위치
COLUMNS = ["text", "label", "prob", "timestamp"]        # 로그 컬럼 순서


def save_log(df):
    """예측 결과에 예측 시각을 붙여 로그 파일 끝에 추가"""
    LOG_PATH.parent.mkdir(exist_ok=True)                # data/ 폴더가 없으면 만듦
    df = df.assign(timestamp=pd.Timestamp.now())        # 예측 시각 컬럼 추가
    df[COLUMNS].to_csv(LOG_PATH, mode="a", header=False, index=False)  # 헤더 없이 이어 쓰기


def load_log():
    """로그 파일 읽기 (아직 없으면 빈 표)"""
    if not LOG_PATH.exists():                           # 예측을 한 번도 안 했다면
        return pd.DataFrame(columns=COLUMNS)            # 컬럼만 있는 빈 표
    return pd.read_csv(LOG_PATH, names=COLUMNS)         # 헤더가 없으므로 컬럼 이름을 지정
```

### 4.3 예측 페이지 (`pages/03_model_demo.py`)

여러 문장을 한 번에 입력받아 예측하고, 2장에서 배운 컬럼·지표·표·차트로 결과를 보여줍니다.

```python
import time
import streamlit as st
from src.model import predict_texts                     # 4.1에서 만든 예측 함수
from src.utils import save_log                          # 4.2에서 만든 로그 저장 함수

st.header("텍스트 감성 예측")
st.caption("학습한 단어: good, great, awesome, bad, awful, terrible")  # 더미 모델이 아는 단어 안내

texts = st.text_area("문장 입력 (한 줄에 한 문장)", "good\nbad\nawesome")  # 여러 줄 입력

if st.button("예측 실행"):                               # 버튼을 누르면 예측 시작
    input_list = [t for t in texts.split("\n") if t.strip()]  # 줄 단위로 나누고 빈 줄 제거
    try:
        t0 = time.perf_counter()                        # 시작 시각
        df = predict_texts(input_list)                  # 모델 예측
        dt = time.perf_counter() - t0                   # 처리 시간
    except Exception as e:                              # 빈 입력 등으로 예측이 실패하면
        st.error(f"예측 중 오류 발생: {e}")              # 빨간 오류 메시지
        st.stop()                                       # 아래 코드는 실행하지 않음

    col1, col2, col3 = st.columns(3)                    # 지표 3개를 가로로 배치
    col1.metric("입력 문장 수", len(df))
    col2.metric("긍정 판정", int(df["label"].sum()))     # label이 1인 문장 수
    col3.metric("처리 시간", f"{dt:.3f}초")

    st.dataframe(df)                                    # 문장별 예측 결과 표
    st.bar_chart(df.set_index("text")["prob"])          # 문장별 긍정 확률 막대 차트
    save_log(df)                                        # data/log.csv에 누적 저장
    st.toast("예측 완료! 🎉")                            # 화면 구석에 잠깐 뜨는 알림
```

| `st.toast()` | 화면 구석에 잠깐 떴다 사라지는 알림 |
| `st.error()` | 빨간색 오류 메시지 |
| `st.stop()` | 그 자리에서 스크립트 실행 중단 |

Cloud가 앱을 실행하기 전에 **설치할 패키지 목록**입니다. 빠진 패키지가 있으면 `ModuleNotFoundError`가 납니다.
</details>

<details>
<summary>2. API 키를 GitHub에 올리지 않고 Cloud 웹페이지에 등록하는 이유는?</summary>

GitHub에 올린 코드는 다른 사람도 볼 수 있어 키가 노출됩니다. Cloud Secrets에 등록하면 **앱만 읽을 수 있습니다.**
</details>

<details>
<summary>3. 코드를 수정한 뒤 Cloud 앱에 반영하려면?</summary>

`git push`만 하면 됩니다. Cloud가 저장소 변경을 감지해 **자동으로 다시 배포**합니다.
</details>

<details>
<summary>4. 배포 오류가 나면 어디를 확인하는가?</summary>

앱 화면의 **Manage app** 로그를 보고, 원인에 따라 `requirements.txt`나 **Settings → Secrets**를 확인합니다.
</details>

## 8. 정리 및 복습

### 8.1 핵심 개념 요약

| 구분 | 핵심 내용 | 요약 포인트 |
|------|------------|--------------|
| **실행 원리** | rerun | 상호작용마다 스크립트 전체를 다시 실행 |
| **화면 구성** | 입력 위젯, 출력 요소, `st.columns`, `st.sidebar` | 함수 호출이 곧 화면 |
| **상태 관리** | `st.session_state` | rerun·페이지 이동 후에도 값 유지 |
| **캐싱** | `@st.cache_data`, `@st.cache_resource` | 오래 걸리는 계산·모델 로드를 한 번만 |
| **프로젝트 구조** | `app.py`, `pages/`, `src/`, `data/` | 화면과 로직 분리 |
| **모델 연동** | `load_model()`, `predict_texts()` | 입력 → 예측 → 출력 → 로그 |
| **디자인** | `.streamlit/config.toml`, CSS | 테마는 설정 파일, 세부 모양은 CSS |
| **배포** | GitHub + Community Cloud | push하면 자동 재배포, 비밀 값은 Cloud Secrets |

### 8.3 단답형 문제

<details>
<summary>1. <code>@st.cache_data</code>와 <code>@st.cache_resource</code>의 차이를 2줄로 설명하시오.</summary>

- `@st.cache_data`는 **결과값**(DataFrame, list 등)을 저장하고 복사본을 돌려준다.  
- `@st.cache_resource`는 **객체 자체**(모델, DB 연결)를 저장하고 같은 객체를 돌려준다.
</details>

<details>
<summary>2. Streamlit과 FastAPI의 역할 차이를 2줄로 요약하시오.</summary>

- Streamlit은 **사람이 브라우저로 사용하는 화면**을 Python만으로 만드는 프레임워크이다.  
- FastAPI는 **다른 프로그램이 호출하는 API 서버**를 만드는 백엔드 프레임워크이다.
</details>

<details>
<summary>3. Streamlit에서 <code>st.session_state</code>가 필요한 이유는?</summary>

상호작용마다 스크립트 전체가 rerun되어 일반 변수가 초기화되므로, **유지해야 할 값**을 저장하기 위해서이다.
</details>

<details>
<summary>4. Streamlit 앱의 rerun 비용을 줄이는 방법은?</summary>

오래 걸리는 데이터 로드·연산에는 `@st.cache_data`, 모델 로드에는 `@st.cache_resource`를 붙여 **한 번만 실행**되게 한다.
</details>
